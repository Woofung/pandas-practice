```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)                       # 固定亂數種子，每次結果都一樣
genes = [f"GENE{i:04d}" for i in range(1, 1001)]      # GENE0001 ~ GENE1000
samples = ["ctrl_1", "ctrl_2", "ctrl_3", "trt_1", "trt_2", "trt_3"]

counts_sim = pd.DataFrame(
    rng.negative_binomial(n=5, p=0.05, size=(1000, 6)),
    index=genes, columns=samples,
)
counts_sim.index.name = "gene_id"
counts_sim.to_csv("../data/counts.csv")

meta_sim = pd.DataFrame({
    "sample_id": samples,
    "condition": ["control"] * 3 + ["treated"] * 3,
    "batch": ["A", "B", "A", "B", "A", "B"],
})
meta_sim.to_csv("../data/metadata.csv", index=False)
```

```python
counts = pd.read_csv("../data/counts.csv", index_col="gene_id")
```
